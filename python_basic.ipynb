 # **Nested Loop**
```python
vehicles = ['car', 'bike']
colors = ['red', 'blue']
for vehicle in vehicles:
  for color in colors:
    print(color, vehicle)
```
```
red car
blue car
red bike
blue bike
```
```python
x=[[1,2,3],["a","b","c"]]
for i in x:
  for j in i:
    print(j,end="")
  print()
```
